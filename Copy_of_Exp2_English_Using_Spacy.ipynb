```python
import spacy
from collections import Counter

# Load English model
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
The boys are playing football in the ground.
The girls are watching the game.
They are enjoying the match.
"""
```

```python
doc = nlp(text)
```

```python
# ---------------------------------------
# 1. Sentence Detection
# ---------------------------------------

print("SENTENCE DETECTION")

for sentence in doc.sents:
    print(sentence.text)
```

```python
# ---------------------------------------
# 2. Tokenization
# ---------------------------------------

tokens = [token.text for token in doc]

print("\nTOKENIZATION")
print(tokens)
```

```python
# ---------------------------------------
# 3. Statistical Analysis
# ---------------------------------------

words = [
    token.text.lower()
    for token in doc
    if token.is_alpha
]

print("\nSTATISTICAL ANALYSIS")

print("Total Tokens:", len(words))
print("Unique Tokens:", len(set(words)))

frequency = Counter(words)

print("\nWord Frequency:")
for word, count in frequency.items():
    print(word, ":", count)
```

```python
# ---------------------------------------
# 4. Lemmatization
# ---------------------------------------

print("\nLEMMATIZATION")

for token in doc:
    if token.is_alpha:
        print(token.text, "->", token.lemma_)
```

```python
# ---------------------------------------
# 5. Stop Word Removal
# ---------------------------------------

print("\nSTOP WORD REMOVAL")

filtered_words = [
    token.text
    for token in doc
    if token.is_alpha and not token.is_stop
]

print(filtered_words)
```
