```python
import stanza
from collections import Counter

nlp = stanza.Pipeline(
    'hi',
    processors='tokenize,pos,lemma'
)

text = """
लड़के मैदान में फुटबॉल खेल रहे हैं।
लड़कियाँ खेल देख रही हैं।
वे मैच का आनंद ले रहे हैं।
"""

doc = nlp(text)

# ---------------------------------------
# Sentence Detection
# ---------------------------------------

print("SENTENCE DETECTION")

for sentence in doc.sentences:
    print(sentence.text)

# ---------------------------------------
# Tokenization
# ---------------------------------------

tokens = []

for sentence in doc.sentences:
    for word in sentence.words:
        tokens.append(word.text)

print("\nTOKENIZATION")
print(tokens)

# ---------------------------------------
# Statistical Analysis
# ---------------------------------------

# Reconstructing words list using Stanza's POS tags to filter out punctuation
words = []
for sentence in doc.sentences:
    for word in sentence.words:
        if word.upos != 'PUNCT': # Filter out punctuation based on UPOS tag
            words.append(word.text.lower())

print("\nSTATISTICAL ANALYSIS")

print("Total Tokens:", len(words))
print("Unique Tokens:", len(set(words)))

frequency = Counter(words)

print("\nWord Frequency:")

for word, count in frequency.items():
    print(word, ":", count)

# ---------------------------------------
# Lemmatization
# ---------------------------------------

print("\nLEMMATIZATION")

for sentence in doc.sentences:
    for word in sentence.words:
        print(word.text, "->", word.lemma)
hindi_stopwords = {
    "में",
    "है",
    "हैं",
    "का",
    "के",
    "की",
    "और",
    "से",
    "को",
    "यह",
    "वे",
    "पर",
    "एक"
}
filtered_words = [
    word for word in words
    if word not in hindi_stopwords
]

print("\nSTOP WORD REMOVAL")
print(filtered_words)
```
