```python
import stanza
from collections import Counter

# ---------------------------------------
# Load Marathi NLP Pipeline
# ---------------------------------------

nlp = stanza.Pipeline(
    "mr",
    processors="tokenize,pos,lemma"
)

# ---------------------------------------
# Marathi Text
# ---------------------------------------

text = """
कृत्रिम बुद्धिमत्ता आज आपल्या जीवनात वेगाने बदल घडवून आणत आहे.
विद्यार्थी वेगवेगळ्या विषयांचा अभ्यास करण्यासाठी संगणकाचा वापर करतात.
शिक्षक कठीण संकल्पना समजावून सांगण्यासाठी तंत्रज्ञानाचा वापर करतात.
नैसर्गिक भाषा प्रक्रिया संगणकाला मानवी भाषा समजण्यास मदत करते.
मशीन लर्निंग मॉडेल मोठ्या प्रमाणातील डेटामधून नमुने शिकू शकतात.
एनएलपी अनुप्रयोग विकसित करण्यासाठी पायथनचा मोठ्या प्रमाणावर वापर केला जातो.
अनेक संशोधक अधिक चांगली भाषा मॉडेल्स तयार करण्यासाठी काम करत आहेत.
ही तंत्रज्ञाने आपल्या दैनंदिन जीवनात महत्त्वाची होत आहेत.
"""

doc = nlp(text)

# ---------------------------------------
# 1. Sentence Detection
# ---------------------------------------

print("SENTENCE DETECTION")

for sentence in doc.sentences:
    print(sentence.text)

# ---------------------------------------
# 2. Tokenization
# ---------------------------------------

tokens = []

for sentence in doc.sentences:
    for word in sentence.words:
        tokens.append(word.text)

print("\nTOKENIZATION")
print(tokens)

# ---------------------------------------
# 3. Statistical Analysis
# ---------------------------------------

words = []
for sentence in doc.sentences:
    for word_obj in sentence.words:
        if word_obj.upos != 'PUNCT':
            words.append(word_obj.text)

print("\nSTATISTICAL ANALYSIS")

print("Total Tokens:", len(words))
print("Unique Tokens:", len(set(words)))

frequency = Counter(words)

print("\nWord Frequency:")

for word, count in frequency.items():
    print(word, ":", count)

# ---------------------------------------
# 4. Lemmatization
# ---------------------------------------

print("\nLEMMATIZATION")

for sentence in doc.sentences:
    for word in sentence.words:
        print(word.text, "->", word.lemma)

# ---------------------------------------
# 5. Stop Word Removal
# ---------------------------------------

marathi_stopwords = {
    "आहे",
    "आहेत",
    "आणि",
    "मध्ये",
    "चा",
    "ची",
    "चे",
    "ला",
    "ने",
    "ते",
    "त्या",
    "सर्व",
    "हे"
}

filtered_words = [
    word for word in words
    if word not in marathi_stopwords
]

print("\nSTOP WORD REMOVAL")

print(filtered_words)
```
