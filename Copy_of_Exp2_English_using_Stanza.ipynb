```python
import stanza
from collections import Counter

# Create pipeline
nlp = stanza.Pipeline(
    'en',
    processors='tokenize,pos,lemma'
)

text = """
The boys are playing football in the ground.
The girls are watching the game.
They are enjoying the match.
"""

doc = nlp(text)
```

```python
# ---------------------------------------
# 1. Sentence Detection
# ---------------------------------------

print("SENTENCE DETECTION")

for sentence in doc.sentences:
    print(sentence.text)
```

```python
# ---------------------------------------
# 2. Tokenization
# ---------------------------------------

tokens = []

for sentence in doc.sentences:
    for word in sentence.words:
        tokens.append(word.text)

print("\nTOKENIZATION")
print(tokens)
```

```python
# ---------------------------------------
# 3. Statistical Analysis
# ---------------------------------------

words = [
    word.lower()
    for word in tokens
    if word.isalpha()
]

print("\nSTATISTICAL ANALYSIS")

print("Total Tokens:", len(words))
print("Unique Tokens:", len(set(words)))

frequency = Counter(words)

print("\nWord Frequency:")
for word, count in frequency.items():
    print(word, ":", count)
```

```python
# ---------------------------------------
# 4. Lemmatization
# ---------------------------------------

print("\nLEMMATIZATION")

for sentence in doc.sentences:
    for word in sentence.words:
        print(word.text, "->", word.lemma)
```

```python
stop_words = {
    "the", "is", "are", "in", "on",
    "a", "an", "and", "of", "to"
}

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("\nSTOP WORD REMOVAL")
print(filtered_words)
```
