```python
import nltk
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer
porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer('english')
```

```python
words = [
    "connect",
    "connected",
    "connection",
    "connecting",
    "compute",
    "computed",
    "computer",
    "computing",
    "relational",
    "relation",
    "relationships",
    "organizing",
    "organization",
    "organized",
    "generously",
    "generation",
    "generative",
    "happiness",
    "nationalization",
    "traditional"
]
```

```python
print("Word\t\tPorter\t\tLancaster\tSnowball")
print("-" * 60)
for word in words:
  porter_stem=porter.stem(word)
  lancaster_stem=lancaster.stem(word)
  snowball_stem=snowball.stem(word)
  print(f"{word:<16}{porter_stem:<16}{lancaster_stem:<16}{snowball_stem}")
```
