**MORPHOLOGICAL PARSER**

REGULAR NOUNS LEXICON




```python
regular_nouns={
    "bureau": "BUREAU",
    "bureaus": "BUREAU",

    "portfolio": "PORTFOLIO",
    "portfolios": "PORTFOLIO",

    "volcano": "VOLCANO",
    "volcanoes": "VOLCANO",

    "echo": "ECHO",
    "echoes": "ECHO",

    "mosquito": "MOSQUITO",
    "mosquitoes": "MOSQUITO",

    "city": "CITY",
    "cities": "CITY",

    "baby":"BABY",
    "babies":"BABY",

    "story":"STORY",
    "stories":"STORY"
}
```

IRREGULAR NOUNS LEXICON

```python
irregular_nouns ={
    "analysis": "ANALYSIS",
    "analyses": "ANALYSIS",

    "thesis": "THESIS",
    "theses": "THESIS",

    "criterion": "CRITERION",
    "criteria": "CRITERION",

    "phenomenon": "PHENOMENON",
    "phenomena": "PHENOMENON",

    "diagnosis": "DIAGNOSIS",
    "diagnoses": "DIAGNOSIS",

    "men": "MAN",
    "man":"MAN",

    "women": "WOMAN",
    "woman": "WOMAN",

    "children": "CHILD",
    "child": "CHILD",

    "mice": "MOUSE",
    "mouse": "MOUSE",

    "geese": "GOOSE",
    "goose": "GOOSE",

    "teeth": "TOOTH",
    "tooth": "TOOTH",

    "feet": "FOOT",
    "foot": "FOOT",

    "people": "PERSON",
    "person": "PERSON",

    "oxen": "OX",
    "ox": "OX"
}
```

METHOD FOR FST PARSER

```python
def fst_parser(word):
  word = word.lower()
  # regular nouns
  if word in regular_nouns:
    root= regular_nouns[word]
    # plural ending in s
    if word.endswith('s') and not word.endswith('es'):
      intermediate=root + " ^ S #"
      number = "PL"

    # plural ending in es
    elif word.endswith('es') and not word.endswith('ies'):
      intermediate=root + " ^ ES #"
      number = "PL"

    # plural ending in ies
    elif word.endswith('ies'):
      intermediate=root + " ^ IES #"
      number = "PL"

    # singular
    else:
      intermediate = root + " #"
      number = "SG"


  # irreglar nouns
  elif word in irregular_nouns:
    root =irregular_nouns[word]
    # singular
    if word == root.lower():
      intermediate = root + " #"
      number = "SG"

    # plural
    else:
      intermediate = root + " #"
      number ="PL"


  # Not found
  else:
    print("Word not found in lexicon")
    return
  output =" "+ root + " + N + "+number
  print("INPUT: ",word.upper())
  print("INTERMEDIATE: ",intermediate)
  print("OUTPUT: ",output)
```

MAIN PROGRAM

```python
print("*"*50)
print(" FST MORPHOLOGICAL PARSER FOR NOUN INFLECTION")
print("*"*50)
while(True):
  word = input("Enter word or type exit to stop: ")
  if word.lower() == "exit":
    break
  fst_parser(word)
  print("-"*50)
```
