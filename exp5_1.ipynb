```python
import nltk
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer


porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer("english")


while True:
    print("\n" + "=" * 45)
    print("        STEMMING PROGRAM")
    print("=" * 45)
    print("1. Enter a word and stem it")
    print("2. Exit")
    print("=" * 45)

    choice = input("Enter your choice: ")

    if choice == "1":
        word = input("\nEnter a word: ")

        porter_stem = porter.stem(word)
        lancaster_stem = lancaster.stem(word)
        snowball_stem = snowball.stem(word)

        print("\nStemming Results")
        print("-" * 40)
        print(f"Original Word  : {word}")
        print(f"Porter Stem    : {porter_stem}")
        print(f"Lancaster Stem : {lancaster_stem}")
        print(f"Snowball Stem  : {snowball_stem}")

    elif choice == "2":
        print("\nExited program")
        break

    else:
        print("\nInvalid choice! Please enter 1 or 2.")
```
