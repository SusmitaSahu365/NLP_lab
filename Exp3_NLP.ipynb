```python
def dfa(word):
  state='q0'
  for ch in word:
    if state=='q0':
      if ch=='b':
        state='q1'
      else:
        return False
    elif state=='q1':
      if ch=='a':
        state='q2'
      else:
        return False
    elif state=='q2':
      if ch=='a':
        state='q3'
      else:
        return False
    elif state=='q3':
      if ch=='a':
        state='q3'
      else:
        return False

  return state=='q3'

print("*"*50)
print(f"{'Word Recognizer':^50}")
print('*'*50)
print()
word=input("Enter a word:")
if(dfa(word)):
  print("Accepted")
else:
  print("Rejected")
```
